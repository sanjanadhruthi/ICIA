```python
from pathlib import Path
import pandas as pd

RAW = Path("../data/raw")

for f in sorted(RAW.rglob("*")):
    if f.suffix not in {".xlsx", ".xls", ".csv"}:
        continue
    df = pd.read_csv(f) if f.suffix == ".csv" else pd.read_excel(f)
    print(f"\n=== {f.parent.name} / {f.name}")
    print("shape:", df.shape)
    print("columns:", list(df.columns))
```

```python
out = []
for f in sorted(RAW.rglob("*")):
    if f.suffix not in {".xlsx", ".xls", ".csv"}:
        continue
    if f.suffix == ".csv":
        df, sheets = pd.read_csv(f, header=None), ["csv"]
    else:
        sheets = pd.ExcelFile(f).sheet_names
        df = pd.read_excel(f, header=None)
    out.append(f"\n===== {f.parent.name} / {f.name} | shape {df.shape} | sheets {sheets}")
    out.append(df.head(8).to_string())
    out.append("...")
    out.append(df.tail(3).to_string())

Path("../data/inventory.txt").write_text("\n".join(out), encoding="utf-8")
print("saved")
```
